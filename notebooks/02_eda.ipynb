<h1>Feature Correlations with Target (Cover_Type)</h1>

Correlation analysis was performed to examine the relationship between the input features and the target variable, `Cover_Type`. This helps identify features that show stronger or weaker linear associations with the target and provides useful insights for subsequent feature analysis and modeling.


```python
sns.heatmap(data=df[["Elevation",
    "Aspect",
    "Slope",
    "Horizontal_Distance_To_Hydrology",
    "Vertical_Distance_To_Hydrology",
    "Horizontal_Distance_To_Roadways",
    "Hillshade_9am",
    "Hillshade_Noon",
    "Hillshade_3pm",
    "Horizontal_Distance_To_Fire_Points",]].corr(),annot=True)
```

```python
df.corr()['Cover_Type'][:-1].sort_values()
```

<h4>Outlier Analysis<h4/>

Box plots were used to identify potential outliers in the numerical features. However, in the Covertype dataset, extreme values are not necessarily errors or missing values, as features such as elevation, distances, and hillshade naturally have different ranges. Therefore, these observations were retained unless there was evidence that they represented invalid data.
